# Exploratory Data Analysis (EDA) - Crop Yield Dataset

This notebook provides a basic analysis of the dataset columns and target behaviors.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv('../data/crop_yield.csv')
print(f"Dataset Shape: {df.shape}")
df.head()
```

```python
# Crop label counts
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='Crop', palette='crest')
plt.title('Distribution of Crop Categories')
plt.ylabel('Count')
plt.xlabel('Crop')
plt.show()
```

```python
# Yield distribution
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x='Yield_tons_per_hectare', kde=True, bins=30, color='forestgreen')
plt.title('Distribution of Yield (tons/ha)')
plt.show()
```

```python
# Average yield by crop
df.groupby('Crop')['Yield_tons_per_hectare'].mean().sort_values(ascending=False)
```
